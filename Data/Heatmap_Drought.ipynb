```python
import pandas as pd

heatmap = pd.read_csv(r"D:\#5 years of PhD\Academics\National Climate Stack Innovation Challenge\Data Sets\#State Wise Data\#LSTM TEST\##Climate-Forecast-main\Data\MAIN_DATA.csv")

```

```python
def heatmap_drought(year):
    # Filter data for Telangana and West Bengal in the specified year
    df_filtered = heatmap[(heatmap['State'].isin(['Telangana', 'West Bengal'])) & (heatmap['year'] == year)]

    # Count 'Severe Drought' and 'Moderate Drought' classes by State and month
    drought_counts = df_filtered[df_filtered['Drought_Class'].isin(['Severe drought', 'Moderate drought'])].groupby(['State', 'month', 'Drought_Class']).size().unstack(fill_value=0)

    # Reset index to make it a cleaner dataframe
    result_df = drought_counts.reset_index()
    print(result_df)



for year in range(2013, 2025):
    print("======================================================================")
    print(f"\t\t\t\tYear: {year}")
    print("======================================================================")

    heatmap_drought(year)
```
